# Space (Inter-Satellite) Channel Simulation


```python
import sys
sys.path.append('../')
from owcm import Channel
import numpy as np
import matplotlib.pyplot as plt
```

## Free Space Path Loss


```python
d_km = np.logspace(2, 5, 20) # 100km to 100,000km
raw_losses = []
for d in d_km:
    ch = Channel.environment('space').defaults('deep_space').build(
        distance_m=d*1000, wavelength_nm=1550
    )
    raw_losses.append(10*np.log10(ch.gain()))

plt.semilogx(d_km, raw_losses)
plt.xlabel('Distance (km)')
plt.ylabel('Path Loss (dB)')
plt.title('FSPL for Intersatellite Link')
plt.grid(True)
plt.show()
```
