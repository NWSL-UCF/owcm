# Indoor (VLC) Channel Simulation


```python
import sys
sys.path.append('../')
from owcm import Channel
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Gain vs Distance (Lambertian)
Plotting the DC channel gain as we move away from the source.


```python
distances = np.linspace(1, 5, 50)
gains = []
for d in distances:
    c = Channel.environment('indoor').defaults('los').build(
        distance_m=d, lambertian_order=1.0, DR_m=0.01
    )
    gains.append(c.gain())

plt.plot(distances, gains, 'g-o')
plt.xlabel('Distance (m)')
plt.ylabel('DC Gain')
plt.title('Indoor Lambertian LoS Gain')
plt.grid(True)
plt.show()
```
