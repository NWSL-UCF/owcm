# Underwater Channel Simulation


```python
import sys
sys.path.append('../')
from owcm import Channel
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Attenuation vs Turbidity
Comparing Clear Ocean vs Turbid Water.


```python
R = 20.0 # meters
extinction_coeffs = np.linspace(0.05, 0.5, 50) # From clear to turbid
losses = []
for c in extinction_coeffs:
    ch = Channel.environment('underwater').defaults('beer_lambert').build(
        distance_m=R, attenuation_c=c
    )
    losses.append(10 * np.log10(ch.gain()))

plt.plot(extinction_coeffs, losses)
plt.xlabel('Extinction $c$ (1/m)')
plt.ylabel('Loss (dB)')
plt.title(f'Underwater Loss at R={R}m')
plt.grid(True)
plt.show()
```
